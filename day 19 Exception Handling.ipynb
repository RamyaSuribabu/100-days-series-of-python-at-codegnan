```python
# execute the corresponding except block code
try:
    a=10
    b=0
    c=a/b
    print(c)
except ZeroDivisionError:
    print("zero division not possible")
except:
    print("something wrong in logic")  
```

```python
# multiple exceptions in single except statement
try:
    a=[10,20,30]
    c=a[10]
    print(c)
except (ZeroDivisionError,IndexError):
    print("zero division not possible")
except TypeError:
    print("we connot add integer with  string")
except Exception as e :
    print(e)
```
