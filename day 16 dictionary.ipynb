#Dictonary - it is ordered collection of key-value pair.it is mutable and represented with '{}'
#properties -->
-->key should be unique
-->values allows duplicates
-->list cannot be used as a key
-->all set elements can be used as dictonary keys
-->dictonary is mutable
-->dictonary is not index based  and not slicable
-->dictonary values access by using key's
EX: syntax: {key : value}  <------------called as item
--> key and value is separated with colon(:)
EX: { 'name' : 'ramya','age':50}
-->Items are separate d with comma(,)

```python
details={ 'name' : 'ramya','age' : 21, 'batch' : 47,'course' : 'python'}
print(type(details))
print(details)

```

```python
# access dictonary values
details={ 'name' : 'ramya','age' : 21, 'batch' : 47,'course' : 'python'}
print(details['name'])
print(details['age'])             # if key is not present in the dictonary then it gives key error
print(details['course'])





```

```python
#updating the values and items
details={ 'name' : 'ramya','age' : 21, 'batch' : 47,'course' : 'python'}
details['name']='babu'
print(details)
# assigning new item into the dictonary   
#_________________________________________
#dict_object[key]= value
#_________________________________________
details['ash'] = 'raf'
print(details)
```

```python
## dictinary functions
# accessing values
# get(key,default_value) : it returns the= value if key is present in dictonary otherwise it returns default value
details={ 'name' : 'ramya','age' : 21, 'batch' : 47,'course' : 'python'}
print(details.get('name'))
print(details.get('percentage'))   #the key is not present in the dictonary so it returns  "none"
print(details.get('percantage',90))


```

```python
#adding or updating  items to dictionary
#update(dictonary) : it updates with new dictonary items
details={ 'name' : 'ramya','age' : 21, 'batch' : 47,'course' : 'python','percentage' : 0}
marks={'python' :90,'soft skills'  :35,'aptitude' : 78,'percentage' : 90}
details.update(marks)
print(details)
print(marks)



```

```python
# dictionary not allow duplicate keys
details={ 'name' : 'ramya','age' : 21, 'batch' : 47,'course' : 'python','percentage' : 0,'percentage' : 90}
print(details)
```

```python
#get keys and values
# keys() - it returns list of all dictonary keys
# values() - it returns list of all dictonary value
# items () - it returns all dictonary items as list of tuple of key-value pairs
details={ 'name' : 'ramya','age' : 21, 'batch' : 47,'course' : 'python','percentage' : 0,'percentage' : 90}
print(details.keys())
print(details.values())
print(details.items())
print(type(details.keys()))
print(type(details.values()))
print(type(details.items()))
print(list(details.keys()))
```

```python
#deleting items from dictonary
#pop(key) - it return value and removes the specified key item from dictonary
#popitem() - it return and removes the last item from dictonary
details={ 'name' : 'ramya','age' : 21, 'batch' : 47,'course' : 'python','percentage' : 90}
poped_value = details.pop('age')
print(poped_value)
print(details)
poped_value = details.pop('age', None)
print(poped_value)
print(details)
item = details.popitem()
print(item)
print(details)
```

```python
details={ 'name' : 'ramya','age' : 21, 'batch' : 47,'course' : 'python','percentage' : 90,'name' : 'ramya'}
print(len(details))
```

```python
#iterate dictonary items
details={ 'name' : 'ramya','age' : 21, 'batch' : 47,'course' : 'python','percentage' : 90,'name' : 'ramya'}
for item in details.items():
    print(item)
```

```python
#iterate dictonary items
details={ 'name' : 'ramya','age' : 21, 'batch' : 47,'course' : 'python','percentage' : 90,'name' : 'ramya'}
for item in details.items():
    print(f"{item[0]}:{item[1]}")
```

```python
key,value= { "asd" : 90}
print(key,value)
```

```python
#iterate dictonary key,value
details={ 'name' : 'ramya','age' : 21, 'batch' : 47,'course' : 'python','percentage' : 90,'name' : 'ramya'}
for key,value  in details.items():
    print(f"{key}:{value}")
```

```python
#membership checking
details={ 'name' : 'ramya','age' : 21, 'batch' : 47,'course' : 'python','percentage' : 90}
if 'name' in details :
    print("key is present")
else:
    print("key is not present")
    
```

```python
#membership checking
details={ 'name' : 'ramya','age' : 21, 'batch' : 47,'course' : 'python','percentage' : 90}
if 'python' in details.values() :
    print("value is present")
else:
    print("value is not present")
```
