```python
book management using set & dictionary
set- it stores book reg no
dict- it stores book details
{101 :{'name' : 'python','author':'as'}}
{101:['python' :'as']}
operations   |key              |dict    
1.add book - id,      name,author,quantity
2.delete book - id id - delete from set
                   book - dlete from dict
3.total books -len()
4.allbooks
5.barrow book - 
```

```python
# Set to store book IDs
book_ids = set()

# Dictionary to store book details
books = {}

# 1. Add Book
def add_book():
    book_id = int(input("Enter Book ID: "))
    
    if book_id in book_ids:
        print("Book already exists!")
        return
    
    name = input("Enter Book Name: ")
    author = input("Enter Author Name: ")
    quantity = int(input("Enter Quantity: "))
    
    book_ids.add(book_id)
    books[book_id] = {
        'name': name,
        'author': author,
        'quantity': quantity
    }
    print("Book added successfully ")

# 2. Delete Book
def delete_book():
    book_id = int(input("Enter Book ID to delete: "))
    
    if book_id in book_ids:
        book_ids.remove(book_id)
        del books[book_id]
        print("Book deleted successfully ")
    else:
        print("Book not found!")

# 3. Total Books
def total_books():
    print("Total number of books:", len(book_ids))

# 4. Display All Books
def display_books():
    if not books:
        print("No books available.")
        return
    
    for book_id, details in books.items():
        print("\nBook ID:", book_id)
        print("Name:", details['name'])
        print("Author:", details['author'])
        print("Quantity:", details['quantity'])

# 5. Borrow Book
def borrow_book():
    book_id = int(input("Enter Book ID to borrow: "))
    
    if book_id in books:
        if books[book_id]['quantity'] > 0:
            books[book_id]['quantity'] -= 1
            print("Book borrowed successfully ")
        else:
            print("Book out of stock!")
    else:
        print("Book not found!")

# Main Program
def main():
    while True:
        print("\n--- Book Management System ---")
        print("1. Add Book")
        print("2. Delete Book")
        print("3. Total Books")
        print("4. Display All Books")
        print("5. Borrow Book")
        print("6. Exit")
        
        choice = int(input("Enter your choice: "))
        
        if choice == 1:
            add_book()
        elif choice == 2:
            delete_book()
        elif choice == 3:
            total_books()
        elif choice == 4:
            display_books()
        elif choice == 5:
            borrow_book()
        elif choice == 6:
            print("Thank you! ")
            break
        else:
            print("Invalid choice!")
main()
```
